## Embeddings

```python
mfile = '/s/project/geno2pheno/funcrvp/paper_revisions/supplementary_results/missingness/plof_gene_list.txt'
with open(mfile, 'r') as f:
    plof_gene_list = [line.strip() for line in f.readlines()]

mfile = '/s/project/geno2pheno/funcrvp/paper_revisions/supplementary_results/missingness/deeprvat_gene_list.txt'
with open(mfile, 'r') as f:
    da_gene_list = [line.strip() for line in f.readlines()]

print(len(da_gene_list), len(plof_gene_list))
```

```python
emb_dir = '/s/project/geno2pheno/data/embeddings'

emb_list = []
emb_genes = []
da_emb_genes = []
plof_emb_genes = []
skip_list = []
for emb_file in os.listdir(emb_dir):
    if emb_file.endswith('.tsv'):
        try:
            emb_list.append(emb_file[:-4])
            eg = set(pd.read_csv(f'{emb_dir}/{emb_file}', sep='\t').gene_id)
            emb_genes.append(len(eg))
            da_emb_genes.append(len(eg.intersection(set(da_gene_list))))
            plof_emb_genes.append(len(eg.intersection(set(plof_gene_list))))
        except:
            skip_list.append(emb_file)

emb_genes_df = pd.DataFrame({
    'emb': emb_list,
    'emb_genes': emb_genes,
    'deeprvat_emb': da_emb_genes,
    'plof_gene_list': plof_emb_genes
})

emb_genes_df
```
