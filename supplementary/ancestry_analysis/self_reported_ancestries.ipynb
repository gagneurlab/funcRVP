```python
import pandas as pd
import numpy as np
from plotnine import *
```

```python
(
    ggplot(covs, aes(x='genetic_PC_1', y='genetic_PC_2', color='genetic_ancestry')) +
    geom_point(alpha=0.1) +
    theme_classic() +
    theme(figure_size=(8, 7))
)
```

## Self reported ancestry vs genetic ancestry

```python
meta_df = pd.read_parquet('/s/project/uk_biobank/processed/phenotypes/merged.meta.parquet')

data_path = meta_df[(meta_df['field.html']=='22006-0.0') & (meta_df['latest']==True)]['data_path'].values[0]
sr_anc_cols = meta_df[(meta_df['field.showcase']=='21000') & (meta_df['latest']==True)]['col.name']
sr_anc_df = pd.read_parquet(data_path)
sr_anc_df = sr_anc_df[['eid'] + list(sr_anc_cols)]#.set_index('eid')
sr_anc_df
```

```python
id_map = pd.read_parquet('/s/project/uk_biobank/clean/gagnuer_stegle_fam_mapping_with_missing_dummies.parquet')
sr_anc_df = id_map.merge(sr_anc_df, left_on='gagneur_eid', right_on='eid', how='inner').drop(columns=['eid', 'gagneur_eid']).rename(columns={'stegle_eid': 'eid'})
sr_anc_df
```

```python
covs_plt = covs.merge(sr_anc_df, left_on='IID', right_on='eid')
covs_plt
```

```python
covs_plt['ethnic_background_f21000_0_0'].value_counts()
```

```python
covs_plt[covs_plt['genetic_ancestry'] != 'Caucasian']['ethnic_background_f21000_0_0'].value_counts()
```

```python
(
    ggplot(covs_plt[covs_plt['genetic_ancestry'] != 'Caucasian'], 
           aes(x='genetic_PC_1', y='genetic_PC_2', color='ethnic_background_f21000_0_0')) +
    geom_point(alpha=0.5) +
    theme_classic() +
    theme(figure_size=(10, 8))
)
```
