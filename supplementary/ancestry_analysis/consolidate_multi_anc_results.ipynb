# Make plots

## Computing precision and MAD

```python
max_rank = 5000
ip_cols = ['model', 'trait', 'gene_id', 'rank', 'beta']
ip = pd.concat([fpr[ip_cols].query("rank <= @max_rank"), rvat[ip_cols].query("rank <= @max_rank")])
ip
```

```python
op_cols = [ 'trait', 'gene_id', 'beta_test', 'ancestry']
op = pd.concat([ma_rvat[op_cols], rvat_test[op_cols]])
op
```

```python
plt_df = ip.merge(op, on=['trait', 'gene_id'], how='left')
plt_df['beta_dir'] = (plt_df['beta']*plt_df['beta_test'] > 0).astype(int)
plt_df['abs_dev'] = (plt_df['beta'] - plt_df['beta_test']).abs()
plt_df
```

```python
cumulative_median_series = plt_df.groupby(['model', 'ancestry'])['abs_dev'].apply(lambda x: x.expanding().median())
plt_df['cumulative_median_abs_dev'] = cumulative_median_series.reset_index(level=['model', 'ancestry'], drop=True)

cumulative_beta_prop = plt_df.groupby(['model', 'ancestry'])['beta_dir'].apply(lambda x: x.expanding().mean())
plt_df['cumulative_beta_prop'] = cumulative_beta_prop.reset_index(level=['model', 'ancestry'], drop=True)
plt_df
```

```python
plt_df[(plt_df['ancestry']=='EUR (10k)') & (plt_df['model']=='FuncRVP')].sort_values(by='rank').head(20)
```

```python
all_jackknife_results = []
unique_traits = plt_df['trait'].unique()

for excluded_trait in tqdm(unique_traits, desc="Jackknife Iterations"):
    # Create a subset of the dataframe excluding the current trait
    jackknife_df = plt_df[plt_df['trait'] != excluded_trait].copy()

    # Recompute rank within each model
    jackknife_df['abs_beta'] = jackknife_df['beta'].abs()
    jackknife_df = jackknife_df.sort_values(by=['model', 'rank'], ascending=[True, True])
    # jackknife_df = jackknife_df.sort_values(by=['model', 'rank', 'abs_beta'], ascending=[True, True, True])
    jackknife_df['trait_rank'] = jackknife_df.groupby(['model'])['rank'].rank(method='first', ascending=True)
    jackknife_df['beta_dir'] = (jackknife_df['beta']*jackknife_df['beta_test'] > 0).astype(int)

    # Calculate cumulative median absolute deviation
    cumulative_median_series = jackknife_df.groupby(['model', 'ancestry'])['abs_dev'].apply(lambda x: x.expanding().median())
    jackknife_df['cumulative_median_abs_dev'] = cumulative_median_series.reset_index(level=['model', 'ancestry'], drop=True)

    # Calculate cumulative sign precision
    cumulative_beta_prop = jackknife_df.groupby(['model', 'ancestry'])['beta_dir'].apply(lambda x: x.expanding().mean())
    jackknife_df['cumulative_beta_prop'] = cumulative_beta_prop.reset_index(level=['model', 'ancestry'], drop=True)

    # Store the results, adding a column to identify the excluded trait
    jackknife_df['excluded_trait'] = excluded_trait
    all_jackknife_results.append(jackknife_df)

# Concatenate all jackknife results into a single dataframe
jackknife_results_df = pd.concat(all_jackknife_results, ignore_index=True)
jackknife_results_df
```

## Sign precision plots

```python
color_dict = {
    'Burden Test': "#FF7F00",
    'FuncRVP': "#6A3D9A"
}

max_rank = 2000

(
    ggplot(plt_df.query("rank <= @max_rank"), aes(x='rank', y='cumulative_beta_prop')) +
    geom_line(aes(color='model')) +
    geom_vline(aes(xintercept=820), linetype='dashed') +
    geom_vline(aes(xintercept=1114), linetype='dashed') +
    scale_color_manual(color_dict) +
    facet_wrap('ancestry', scales='free') +
    ylab('Sign precision') +
    xlab('Associations ranked by confidence') +
    theme_bw() +
    theme(figure_size=(10, 7))
)
```

```python
def get_percentile_bounds(series, confidence=0.95):
    """
    Helper function to calculate lower and upper percentile bounds for a Series.
    It returns a pandas Series so it can be easily unstacked.
    """
    if series.empty:
        # Handle cases where a group might be empty (though less likely with groupby)
        return pd.Series([np.nan, np.nan], index=['lower_bound', 'upper_bound'])

    lower_percentile = (1 - confidence) / 2 * 100
    upper_percentile = (1 + confidence) / 2 * 100

    lower_bound = np.percentile(series, lower_percentile)
    upper_bound = np.percentile(series, upper_percentile)

    return pd.Series([lower_bound, upper_bound], index=['lower_bound', 'upper_bound'])

# Calculate the confidence intervals using groupby and apply
# This is significantly faster as it avoids explicit Python loops and repeated filtering.
confidence_interval_df = (
    jackknife_results_df.groupby(['rank', 'ancestry', 'model'])['cumulative_beta_prop']
    .apply(get_percentile_bounds)
    .unstack() # This pivots the 'lower_bound'/'upper_bound' from index to columns
    .reset_index() # This converts the grouped columns ('rank', 'ancestry', 'model') back to regular columns
)

print(confidence_interval_df.head())
print(f"\nShape of the resulting DataFrame: {confidence_interval_df.shape}")

jk_plt_df = plt_df.merge(confidence_interval_df, on=['rank', 'ancestry', 'model'])
jk_plt_df
```

```python
color_dict = {
    'Burden Test': "#FF7F00",
    'FuncRVP': "#6A3D9A"
}

max_rank = 2000

(
    ggplot(jk_plt_df.query("rank >= 10").query("rank <= @max_rank"), aes(x='rank', y='cumulative_beta_prop')) +
    geom_line(aes(color='model')) +
    geom_ribbon(aes(fill='model', ymin='lower_bound', ymax='upper_bound'), alpha=0.4) +
    geom_vline(aes(xintercept=820), linetype='dashed') +
    geom_vline(aes(xintercept=1114), linetype='dashed') +
    scale_color_manual(color_dict) +
    scale_fill_manual(color_dict) +
    facet_wrap('ancestry', scales='free') +
    ylab('Median absolute deviation') +
    xlab('Associations ranked by confidence') +
    theme_bw() +
    theme(figure_size=(10, 7))
)
```
